```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

```

```python
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

# a) config_json - Neural network model architecture , number of hidden layer, hidden dimensions , attention head

# b) model.safetensors - Brain of model life containing trained weights and paramaters

# c) loading_weights - transformer library successfully loaded model weishts

# d) generation_config_json - how model should generate text

# e) tokenizer's jsons - for tokening text
```

```python
# Prompt is a method t interact with LLM. Prompt can be text, image, document, etc.
```

```python
prompt = "What is the capital of India considering Paris is capital of France"
```

```python
inputs = tokenizer.encode(prompt, return_tensors = "pt")
print(inputs)
```

```python
embeddings = model.transformer.wte(inputs)
print(embeddings)
```

```python
# Decoding
output = model.generate(inputs, max_length = 20, num_return_sequences = 1, no_repeat_ngram_size = 2, early_stopping = True)
```

```python
output
```

```python
response = tokenizer.decode(output[0], skip_special_tokens = True)
```

```python
response
```

```python
# GPT2 Models are with context models. search for answers from the trained text only
# and context must be provided for these models to perform
# Context means it should get a relevance.
# Context means an example of expected answer must be provided.
# Context window is how much text model should remember at once.
# Smaller context window forgets earlier text.
# Larger context window understands context better as long sequences are memorised.
```

```python
# Without context,  models are trained on large scale data like world wide web and
# memorize their learning during training phase entirely.
# Context model will just resort to the data we provide for training.
```

```python
from transformers import pipeline
```

```python
# Initiate text generation pipeline
generator = pipeline("text-generation", model = "TinyLlama/TinyLlama-1.1B-Chat-v1.0")
```

```python
#Define the prompt
prompt = "Question: What is the capital of India\nAnswer"
```

```python
outputs = generator(prompt,max_new_tokens = 20, do_sample=False)
```

```python
print(outputs[0]['generated_text'])
```
